# Exercise 3 - Self-Reflection Prompt for Improving Output

**Tools used:** ChatGPT and Google Colab (Python)

**Goal:** Ask the AI to critique an original summary against specific requirements and then produce an improved version.

## Source text

Prompt engineering is the practice of writing and refining instructions so an AI system produces useful results. Clear prompts usually state the role, task, context, desired format, and important constraints. More advanced techniques include prompt chaining, where one output becomes the input to another step, and ReACT-style prompting, which combines planning with actions and observations. Testing and revising prompts is important because small wording changes can affect the quality, consistency, and usefulness of an AI response.

## Original prompt

Summarize the source text in 2-3 sentences for a college student.

## Original summary (before)

Prompt engineering means giving AI instructions so it can answer questions better. It can use different techniques, and prompts should be tested and improved.

## Self-reflection prompt

Critique the original summary using these requirements: (1) accuracy - it must reflect the source without adding unsupported facts; (2) clarity - it should be easy for a college student to understand; (3) focus - it should explain what prompt engineering is and mention at least two practices from the source; (4) length - 60 to 80 words; and (5) format - one paragraph with no bullets. First give a brief critique, then write a revised summary that meets every requirement.

## AI critique

The original summary is accurate but too general. It does not explain the main parts of a clear prompt, and it does not name specific techniques such as prompt chaining or ReACT-style prompting. It is also shorter than the required 60-80 words. The revision should keep the language simple while adding the missing details.

## Improved summary (after)

Strong prompts clearly describe the task, context, desired format, and important constraints. Techniques such as prompt chaining connect multiple steps, while ReACT-style prompting combines planning, actions, and observations. Testing and revising prompts is also important because small wording changes can improve the quality of the final response.
