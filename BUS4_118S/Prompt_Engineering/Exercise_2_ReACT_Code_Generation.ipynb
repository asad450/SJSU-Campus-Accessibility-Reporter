# Exercise 2 - Code Generation with ReACT Prompting

**Tools used:** ChatGPT and Google Colab (Python)

**Goal:** Use a ReACT-style cycle to plan, generate, run, observe, and improve Python code.

## Full ReACT prompt

Act as a Python coding assistant. Follow these stages and label them clearly:

1. **PLAN:** Give a short plan for the solution.
2. **ACTION:** Generate the Python code.
3. **RUN:** Execute the code with a test case.
4. **OBSERVATION:** Check the output or error.
5. **FIX:** If the observation shows a weakness or error, revise the code and run it again.

Task: Write a function named `analyze_sales(sales)` that accepts a list of sales numbers and returns the total, average, and highest sale. Use only built-in Python. The final version must handle an empty list and reject non-numeric values with a clear message. Print the results in an easy-to-read format.

## PLAN

Use `sum`, `len`, and `max` for the calculations. First test a basic version. Then test an edge case with an empty list, observe the problem, and add validation in the final version.

## ACTION - Version 1

The first version assumes the list is non-empty. We will run it on an empty list to test that assumption.

```python
def analyze_sales_v1(sales):
    total = sum(sales)
    average = total / len(sales)
    highest = max(sales)
    return total, average, highest

try:
    print(analyze_sales_v1([]))
except Exception as error:
    print("OBSERVATION:", type(error).__name__, "-", error)
```

## OBSERVATION

The empty-list test causes a division-by-zero error. The prompt required basic error handling, so the code needs one improvement cycle before the final run.

## FIX - Final version

Add checks for an empty list and for non-numeric values, then run both a normal test and the empty-list edge case.

```python
def analyze_sales(sales):
    if not sales:
        return "No sales data provided."
    if not all(isinstance(value, (int, float)) for value in sales):
        return "All sales values must be numeric."

    total = sum(sales)
    average = total / len(sales)
    highest = max(sales)
    return f"Total: ${total:.2f} | Average: ${average:.2f} | Highest: ${highest:.2f}"

print("Final run:", analyze_sales([120, 95, 160, 130]))
print("Edge case:", analyze_sales([]))
```
