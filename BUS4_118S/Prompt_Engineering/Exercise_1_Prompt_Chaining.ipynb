# Exercise 1 - Prompt Chaining for a Customer Support AI

**Tools used:** ChatGPT and Google Colab (Python)

**Goal:** Build a simple multi-step customer support flow where each step uses the output from the previous step.

## Prompt used to generate the code

Create simple Python code that simulates a customer-support AI using prompt chaining. Use four connected steps: (1) classify the customer's issue, (2) determine what information is needed based on that classification, (3) propose a concise solution using the customer message plus the earlier outputs, and (4) decide whether the case should be escalated. Each step must use prior output. Use a professional tone, only built-in Python, and display the result from every step.

## Prompts in the chain

**Step 1 - Classify:** Read the customer's message and classify it as billing, technical, account, shipping, or other.

**Step 2 - Gather information:** Using the classification from Step 1, state the minimum additional information needed to help the customer.

**Step 3 - Propose solution:** Using the original customer message, Step 1 classification, and Step 2 information request, provide a short helpful solution.

**Step 4 - Escalation rule:** Using the issue classification and proposed solution, decide whether the case should be escalated to a human agent. Escalate when there is suspected fraud, an unauthorized charge, account security risk, or when the issue cannot be solved with the available information.

Each step below passes its output into the next step.

## Testing and iteration

**Before:** "Help the customer with their problem."

**Problem:** This was too vague and did not guarantee a real prompt chain.

**Improved prompt:** The final prompt above adds four specific stages, issue categories, tone/format constraints, and an escalation rule. This makes each step depend on earlier output.

```python
def classify_issue(message):
    text = message.lower()
    if any(word in text for word in ["charged", "charge", "billing", "refund", "payment"]):
        return "billing"
    if any(word in text for word in ["login", "password", "account"]):
        return "account"
    if any(word in text for word in ["error", "app", "website", "technical"]):
        return "technical"
    if any(word in text for word in ["delivery", "shipping", "package"]):
        return "shipping"
    return "other"


def gather_missing_info(issue):
    requests = {
        "billing": "Ask for the date and amount of the charge plus the last four digits of the payment method.",
        "account": "Ask for the account email and the exact login or account error.",
        "technical": "Ask for the device, browser/app version, and exact error message.",
        "shipping": "Ask for the order number and current tracking status.",
        "other": "Ask for one more sentence describing the problem and what outcome is expected.",
    }
    return requests[issue]


def propose_solution(message, issue, needed_info):
    if issue == "billing":
        return ("Check the two transaction entries to see whether one is pending or duplicated. "
                f"Next information needed: {needed_info} After that, support can verify the duplicate and correct it if needed.")
    if issue == "account":
        return f"Verify the account details first. Next information needed: {needed_info} Then use those details to choose the next account-recovery step."
    if issue == "technical":
        return f"Restart the app or browser and reproduce the issue. Next information needed: {needed_info} Use those details to troubleshoot the error."
    if issue == "shipping":
        return f"Check the shipment first. Next information needed: {needed_info} If tracking has not moved, contact the carrier or support."
    return f"Use the customer's extra details to identify the correct support process. Next information needed: {needed_info}"


def should_escalate(message, issue, solution):
    text = message.lower()
    high_risk = any(term in text for term in ["fraud", "unauthorized", "hacked", "stolen"])
    cannot_resolve = "cannot" in solution.lower() or "unable" in solution.lower()
    return "Yes - send to a human support agent." if high_risk or cannot_resolve else "No - continue with the proposed solution first."


customer_message = "I was charged twice for my monthly subscription."

step1 = classify_issue(customer_message)
step2 = gather_missing_info(step1)
step3 = propose_solution(customer_message, step1, step2)
step4 = should_escalate(customer_message, step1, step3)

print("Customer message:", customer_message)
print("Step 1 - Classification:", step1)
print("Step 2 - Missing information:", step2)
print("Step 3 - Proposed solution:", step3)
print("Step 4 - Escalation:", step4)
```
